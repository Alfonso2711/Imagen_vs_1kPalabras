#LIBRERIAS

```python
import io
import urllib.request
import zlib

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

# CONFIGURACIÓN

```python
IMAGE_SIZE = 1000
GRID_SIZE = 32

IMAGE_URL = "https://picsum.photos/1000/1000"
```

# 1. FUNCIÓN — ENTROPÍA DE SHANNON

```python
def calcular_entropia(datos):
    datos = np.asarray(datos).ravel()

    if len(datos) == 0:
        return 0.0

    _, conteos = np.unique(
        datos,
        return_counts=True
    )

    probabilidades = conteos / len(datos)

    return -np.sum(
        probabilidades *
        np.log2(probabilidades)
    )
```

# 2. FUNCIÓN — MAPA DE ENTROPÍA

```python
def crear_mapa_entropia(datos, grid_size=32):
    datos = np.asarray(datos)

    alto, ancho = datos.shape[:2]
    bloques = []

    alto_bloque = alto // grid_size
    ancho_bloque = ancho // grid_size

    for fila in range(grid_size):

        for columna in range(grid_size):

            y_inicio = fila * alto_bloque
            y_fin = (
                (fila + 1) * alto_bloque
                if fila < grid_size - 1
                else alto
            )

            x_inicio = columna * ancho_bloque
            x_fin = (
                (columna + 1) * ancho_bloque
                if columna < grid_size - 1
                else ancho
            )

            bloque = datos[
                y_inicio:y_fin,
                x_inicio:x_fin
            ]

            bloques.append(
                calcular_entropia(bloque)
            )

    return np.array(bloques).reshape(
        grid_size,
        grid_size
    )
```

# 3. FUNCIÓN — COMPRESIÓN

```python

def comprimir(datos):
    """
    Comprime bytes utilizando zlib.
    """

    comprimido = zlib.compress(
        datos,
        level=9
    )

    return len(comprimido)
```

# 4. CREAR LAS 1000 PALABRAS

```python
texto_base = """
En un lugar de la Mancha de cuyo nombre no quiero acordarme
no ha mucho tiempo que vivia un hidalgo de los de lanza en
astillero adarga antigua rocin flaco y galgo corredor.
Una olla de algo mas vaca que carnero salpicon las mas noches
duelos y quebrantos los sabados lentejas los viernes algun
palomino de añadidura los domingos consumian las tres partes
de su hacienda.
"""

palabras_base = texto_base.split()

# Generamos exactamente 1000 palabras
palabras = [
    palabras_base[i % len(palabras_base)]
    for i in range(1000)
]

texto = " ".join(palabras)


assert len(palabras) == 1000


print("\n")
print(f"Palabras:{len(palabras):>10,}")
print(f"Caracteres:{len(texto):>10,}")
```

# 5. TEXTO A BYTES

```python
datos_texto = texto.encode("utf-8")
bytes_texto = len(datos_texto)
bits_texto = bytes_texto * 8


# Entropia del texto
caracteres_texto = np.array(
    list(texto)
)

entropia_texto = calcular_entropia(
    caracteres_texto
)

# Estimacion de bits segun entropia
bits_teoricos_texto = (
    entropia_texto *
    len(caracteres_texto)
)

# Compresion
bytes_texto_comprimido = comprimir(
    datos_texto
)
```

# 6. CREAR MAPA DE ENTROPÍA DEL TEXTO

Dividimos el texto en exactamente GRID_SIZE por bloques.

Por ejemplo: De 6144 bytes a 1024 bloques a 6 bytes por bloque

```python
# Convertimos el texto a bytes
valores_texto = np.frombuffer(
    datos_texto,
    dtype=np.uint8
)

# Numero total de regiones
total_celdas = GRID_SIZE * GRID_SIZE

bloques_texto = np.array_split(
    valores_texto,
    total_celdas
)


# Entropia de cada bloque
entropias_texto = np.array([
    calcular_entropia(bloque)
    for bloque in bloques_texto
])

# Convertimos los 1024 valores en 32 × 32
mapa_texto = entropias_texto.reshape(
    GRID_SIZE,
    GRID_SIZE
)
```

# 7. DESCARGAR FOTOGRAFÍA

```python
try:

    with urllib.request.urlopen(
        IMAGE_URL
    ) as respuesta:

        contenido = respuesta.read()

    imagen = Image.open(
        io.BytesIO(contenido)
    ).convert("RGB")


except Exception as error:

    print(
        "\nNo se pudo descargar la imagen."
    )

    print(error)

    raise SystemExit


# Aseguramos 1000 × 1000
imagen = imagen.resize(
    (IMAGE_SIZE, IMAGE_SIZE)
)


imagen_array = np.array(
    imagen,
    dtype=np.uint8
)


print(
    f"Dimensiones: "
    f"{imagen_array.shape[0]} × "
    f"{imagen_array.shape[1]} × "
    f"{imagen_array.shape[2]}"
)
```

# 8. CREAR IMAGEN NEGRA

```python
imagen_negra = np.zeros_like(
    imagen_array,
    dtype=np.uint8
)
```

# 9. CONVERTIR A ESCALA DE GRISES

Convierte RGB a escala de grises utilizando luminancia aproximada.

```python
def rgb_a_grises(imagen):
    return (
        0.299 * imagen[:, :, 0]
        + 0.587 * imagen[:, :, 1]
        + 0.114 * imagen[:, :, 2]
    ).astype(np.uint8)


gris_foto = rgb_a_grises(
    imagen_array
)

gris_negra = rgb_a_grises(
    imagen_negra
)
```

# 10. ENTROPÍA GLOBAL

```python
entropia_foto = calcular_entropia(
    gris_foto
)

entropia_negra = calcular_entropia(
    gris_negra
)
```

# 11. MAPAS DE ENTROPÍA

```python
mapa_foto = crear_mapa_entropia(
    gris_foto,
    GRID_SIZE
)


mapa_negra = crear_mapa_entropia(
    gris_negra,
    GRID_SIZE
)
```

# 12. INFORMACIÓN DE LA REPRESENTACIÓN

```python

bytes_imagen = imagen_array.nbytes
bytes_negra = imagen_negra.nbytes
bits_imagen = bytes_imagen * 8

# Estimacion teorica
bits_teoricos_imagen = (
    entropia_foto *
    gris_foto.size
)


bits_teoricos_negra = (
    entropia_negra *
    gris_negra.size
)
```

# 13. COMPRESIÓN

```python
bytes_imagen_comprimida = comprimir(
    imagen_array.tobytes()
)


bytes_negra_comprimida = comprimir(
    imagen_negra.tobytes()
)
```

# 14. RESULTADOS

```python
print("\nTEXTO — 1000 PALABRAS")
print("-" * 75)

print(
    f"Entropía:               "
    f"{entropia_texto:.4f} bits/carácter"
)

print(
    f"Representación UTF-8:   "
    f"{bytes_texto:,} bytes"
)

print(
    f"Bits físicos:           "
    f"{bits_texto:,} bits"
)

print(
    f"Bits estimados Shannon: "
    f"{bits_teoricos_texto:,.0f} bits"
)

print(
    f"Comprimido con zlib:    "
    f"{bytes_texto_comprimido:,} bytes"
)


print("\nIMÁGENES — 1000 × 1000 RGB")
print("-" * 75)

print(
    f"{'Imagen':<18}"
    f"{'Entropía':>15}"
    f"{'Original':>18}"
    f"{'Comprimida':>18}"
)

print("-" * 75)

print(
    f"{'Negra':<18}"
    f"{entropia_negra:>15.4f}"
    f"{bytes_negra:>18,}"
    f"{bytes_negra_comprimida:>18,}"
)

print(
    f"{'Fotografía':<18}"
    f"{entropia_foto:>15.4f}"
    f"{bytes_imagen:>18,}"
    f"{bytes_imagen_comprimida:>18,}"
)
```

# 15. RATIOS DE COMPRESIÓN

```python
ratio_texto = (
    bytes_texto /
    bytes_texto_comprimido
)

ratio_negra = (
    bytes_negra /
    bytes_negra_comprimida
)

ratio_foto = (
    bytes_imagen /
    bytes_imagen_comprimida
)


print("\n")
print("RATIOS DE COMPRESIÓN")
print("-" * 75)

print(
    f"Texto:            {ratio_texto:.2f}x"
)

print(
    f"Imagen negra:     {ratio_negra:.2f}x"
)

print(
    f"Fotografía:       {ratio_foto:.2f}x"
)
```

#CONCLUSIÓN

1. El tamaño de una representación NO determina por sí mismo su cantidad de información.

2. La imagen negra y el ruido tienen exactamente el mismo tamaño sin comprimir.

3. Sin embargo, su entropía es muy diferente.

   Imagen negra:     H = -0.0000
   Fotografía:       H = 7.2611

4. Una señal altamente predecible puede comprimirse mucho más que una señal impredecible.

5. La entropía de Shannon mide incertidumbre estadística, NO significado semántico.
